```python
import random
import csv
from datetime import datetime, timedelta

# Categories for different transaction types
income_categories = [
    "Sales", "Investment Income", "Interest Income", "Revenue", "Consulting Income", 
    "Commission", "Freelance Income", "Rental Income", "Service Revenue"
]

expense_categories = [
    "Entertainment", "Rent", "Employee Salary", "Office Supplies", "Transportation", 
    "Utilities", "Marketing", "Inventory", "Repairs", "Insurance", "Professional Fees"
]

# Transaction types
transaction_types = ["Income", "Expense"]

# Generate data for the entire year 2024
start_date = datetime(2024, 1, 1)
end_date = datetime(2024, 12, 31)

# Create a list to store rows
transactions = []

# Generate daily transactions for 365 days
current_date = start_date
while current_date <= end_date:
    user_id = "CUST006"
    transaction_type = random.choice(transaction_types)
    
    if transaction_type == "Income":
        # Choose a category from the income categories
        category = random.choice(income_categories)
        # Generate a random income amount between 100 and 5000
        amount = round(random.uniform(100, 5000), 2)
    else:
        # Choose a category from the expense categories
        category = random.choice(expense_categories)
        # Generate a random expense amount between 50 and 3000
        amount = round(random.uniform(50, 3000), 2)
    
    # Prepare the row
    row = [user_id, current_date.strftime("%Y-%m-%d"), amount, category, transaction_type]
    transactions.append(row)
    
    # Move to the next day
    current_date += timedelta(days=1)

# Write the data to a CSV file
with open('transactions_CUST006_2024.csv', mode='w', newline='') as file:
    writer = csv.writer(file)
    writer.writerow(["user_id", "date", "amount", "category", "transaction_type"])
    writer.writerows(transactions)

print("CSV file generated successfully!")
```

```python
import random
import csv
from datetime import datetime, timedelta

# Categories for different transaction types
income_categories = [
    "Sales", "Investment Income", "Interest Income", "Revenue", "Consulting Income", 
    "Commission", "Freelance Income", "Rental Income", "Service Revenue"
]

expense_categories = [
    "Entertainment", "Rent", "Employee Salary", "Office Supplies", "Transportation", 
    "Utilities", "Marketing", "Inventory", "Repairs", "Insurance", "Professional Fees"
]

# Transaction types
transaction_types = ["Income", "Expense"]

# Generate data for the entire year 2024
start_date = datetime(2024, 1, 1)
end_date = datetime(2024, 12, 31)

# Create a list to store rows
transactions = []

# Generate daily transactions for 365 days
current_date = start_date
while current_date <= end_date:
    user_id = "CUST007"
    
    # Income transaction
    income_category = random.choice(income_categories)
    income_amount = round(random.uniform(100, 5000), 2)
    income_row = [user_id, current_date.strftime("%Y-%m-%d"), income_amount, income_category, "Income"]
    transactions.append(income_row)

    # Expense transaction
    expense_category = random.choice(expense_categories)
    expense_amount = round(random.uniform(50, 3000), 2)
    expense_row = [user_id, current_date.strftime("%Y-%m-%d"), expense_amount, expense_category, "Expense"]
    transactions.append(expense_row)

    # Move to the next day
    current_date += timedelta(days=1)

# Write the data to a CSV file
with open('transactions_CUST007_2024.csv', mode='w', newline='') as file:
    writer = csv.writer(file)
    writer.writerow(["user_id", "date", "amount", "category", "transaction_type"])
    writer.writerows(transactions)

print("CSV file generated successfully!")
```
